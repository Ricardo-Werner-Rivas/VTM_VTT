### Through exclusive aspects

```python
# Open files
tremere,brujah=PdfReader("sdlc_clan_sheets\\Tremere.pdf"),PdfReader("sdlc_clan_sheets\\Brujah.pdf")

# Look for exclusive aspects (concept or merits/flaws)
fields_tremere,fields_brujah=tremere.get_fields().copy(),brujah.get_fields().copy()
clans=[]
# Try to look for concept
for fields in (fields_tremere,fields_brujah):
    try:
        fields="\n".join(tuple(fields["Concepto"]["/Opt"]))
    # If concept is not within the fields
    except:
        # Look for clan specific merits
        fields="\n".join(tuple(fields["Méritos.0.0"]["/Opt"]))
        clan=fields[fields.find("Méritos"):].split()[1]
        clans.append(clan)
    # If concept is found
    else:
        clan=fields[fields.find("Conceptos"):].split()[1]
        clans.append(clan)

# Transform into tuple (memory-efficient)
clans=tuple(clans)

# Purge memory
del clan,fields

# Show obtained clans
clans
```

### Clan specific *Biblioteca Oscura* character sheets identification functions

#### Lore of the Clans character sheets

```python
# Function for identifying clan specific Lore of the Clans character sheets
def get_clan_sdlc(sdlc_sheet:PdfReader)->str:
    """
    Obtains the character's Clan for a given Clan specific Lore of the Clans character sheet
    
    Arguments
    ---------
    sldc_sheet : `PdfReader`
        Clan specific Lore of the Clans character sheet to identify
    
    Returns
    -------
    str
        Character's Clan
    """
    fields=sdlc_sheet.get_fields()
    try:
        fields="\n".join(tuple(fields["Concepto"]["/Opt"]))
    except:
        fields="\n".join(tuple(fields["Méritos.0.0"]["/Opt"]))
        clan=fields[fields.find("Méritos"):].split()[1]
    else:
        clan=fields[fields.find("Conceptos"):].split()[1]
    return clan
```

```python
# Function testing
from os import listdir
clans=tuple()
for file in listdir("sdlc_clan_sheets"):
    character=PdfReader("sdlc_clan_sheets\\"+file)
    clans=(*clans,get_clan_sdlc(character))
    character.close()
clans
```

#### Basic Clan specific character sheets

```python
# Function for identifying clan specific basic character sheets
def get_clan_basic(basic_sheet:PdfReader)->str:
    """
    Obtains the character's Clan for a given Clan specific basic character sheet
    
    Arguments
    ---------
    basic_sheet : `PdfReader`
        Clan specific basic character sheet to identify
    
    Returns
    -------
    str
        Character's Clan
    """
    basic_sheet_images=tuple(image.image for image in basic_sheet.pages[0].images)
    global clan_images
    for clan,image in clan_images.items():
        if image in basic_sheet_images:
            char_clan=clan
            break
        else:
            char_clan="Setita"
    return char_clan
```
