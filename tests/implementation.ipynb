## Identify *Biblioteca Oscura* clan specific character sheets

### Through images' hashes

```python
# Read character sheets
sdlc_tremere,basic_tremere=PdfReader("Tremere.pdf"),PdfReader("Tremere Editable.pdf")

# Obtain images' representations
sdlc_tremere_hash=repr(sdlc_tremere.pages[0].images[4])
basic_tremere_hash=repr(basic_tremere.pages[0].images[1])

# Obtain text inside the brackets as a list
sdlc_tremere_hash,basic_tremere_hash=(
    # "Saber de los Clanes" character sheet
    sdlc_tremere_hash
    [len("ImageFile("):-1] # Take class name and brackets out
    .split(", ") # Split with the separation ", "
    [-1] # Take the last element
    [len("hash: "):], # Take everything that isn't the image's hash out
    
    # Repeat for basic sheet
    basic_tremere_hash
    [len("ImageFile("):-1]
    .split(", ")
    [-1]
    [len("hash: "):]
)

# Show images' hashes
print(f"Tremere - Básica: {basic_tremere_hash}\nTremere - Saber de los Clanes: {sdlc_tremere_hash}")
```

```python
# Show sdlc clan image
sdlc_tremere.pages[0].images[4].image
```

```python
# Show basic clan image
basic_tremere.pages[0].images[1].image
```

```python
# Close character sheets
sdlc_tremere.close()
basic_tremere.close()
```

**Not identifiable** through images, as images' hashes **change** <ins>upon iteration</ins>.

### Through exclusive aspects

```python
# Read files
tremere,brujah=PdfReader("Tremere.pdf"),PdfReader("Brujah.pdf")

# Look for exclusive aspects (concept or merits/flaws)
fields_tremere,fields_brujah=tremere.get_fields().copy(),brujah.get_fields().copy()
clans=[]
# Try to look for concept
for fields in (fields_tremere,fields_brujah):
    try:
        fields="\n".join(tuple(fields["Concepto"]["/Opt"]))
    # If concept is not within the fields
    except:
        # Look for clan specific merits
        fields="\n".join(tuple(fields["Méritos.0.0"]["/Opt"]))
        clan=fields[fields.find("Méritos"):].split()[1]
        clans.append(clan)
    # If concept is found
    else:
        clan=fields[fields.find("Conceptos"):].split()[1]
        clans.append(clan)

# Transform into tuple (memory-efficient)
clans=tuple(clans)

# Purge memory
del clan,fields

# Show obtained clans
clans
```

```python
# Close files
tremere.close()
brujah.close()
```

**Identifiable** through <ins>clan specific aspects</ins>.
